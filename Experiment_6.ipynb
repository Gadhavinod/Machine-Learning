```python
import numpy as np
import pandas as pd

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import classification_report, accuracy_score, f1_score
```

```python
categories = [
    'comp.graphics',
    'rec.sport.hockey',
    'sci.med',
    'talk.politics.mideast'
]

print("Fetching dataset...")

train_data = fetch_20newsgroups(
    subset='train',
    categories=categories,
    remove=('headers', 'footers', 'quotes')
)

test_data = fetch_20newsgroups(
    subset='test',
    categories=categories,
    remove=('headers', 'footers', 'quotes')
)

y_train = train_data.target
y_test = test_data.target

print(f"Loaded {len(train_data.data)} training documents")
print(f"Loaded {len(test_data.data)} test documents")
```

```python
mnb_vectorizer = CountVectorizer(
    stop_words='english',
    max_features=10000
)

X_train_counts = mnb_vectorizer.fit_transform(train_data.data)
X_test_counts = mnb_vectorizer.transform(test_data.data)

print("Multinomial vectorization completed.")
print("Training data shape:", X_train_counts.shape)
print("Testing data shape:", X_test_counts.shape)
```

```python
bnb_vectorizer = CountVectorizer(
    stop_words='english',
    max_features=10000,
    binary=True
)

X_train_binary = bnb_vectorizer.fit_transform(train_data.data)
X_test_binary = bnb_vectorizer.transform(test_data.data)

print("Bernoulli vectorization completed.")
print("Training data shape:", X_train_binary.shape)
print("Testing data shape:", X_test_binary.shape)
```

```python
mnb = MultinomialNB(alpha=1.0)

mnb.fit(X_train_counts, y_train)

y_pred_mnb = mnb.predict(X_test_counts)

print("Multinomial Naive Bayes training completed.")
```

```python
bnb = BernoulliNB(alpha=1.0)

bnb.fit(X_train_binary, y_train)

y_pred_bnb = bnb.predict(X_test_binary)

print("Bernoulli Naive Bayes training completed.")
```

```python
results = {
    "Multinomial NB": [
        accuracy_score(y_test, y_pred_mnb),
        f1_score(y_test, y_pred_mnb, average='macro')
    ],
    "Bernoulli NB": [
        accuracy_score(y_test, y_pred_bnb),
        f1_score(y_test, y_pred_bnb, average='macro')
    ]
}

df_results = pd.DataFrame(
    results,
    index=["Accuracy", "Macro F1-Score"]
)

print("Naive Bayes Model Comparison")
print("=" * 40)
display(df_results.round(4))
```

```python
print("=" * 25)
print("Multinomial Classification Report")
print("=" * 25)

print(
    classification_report(
        y_test,
        y_pred_mnb,
        target_names=categories
    )
)
```

```python
print("=" * 25)
print("Bernoulli Classification Report")
print("=" * 25)

print(
    classification_report(
        y_test,
        y_pred_bnb,
        target_names=categories
    )
)
```

```python
print("Final Model Comparison")
print("=" * 40)

print(f"Multinomial NB Accuracy : {accuracy_score(y_test, y_pred_mnb):.4f}")
print(f"Multinomial NB Macro F1 : {f1_score(y_test, y_pred_mnb, average='macro'):.4f}")
print()
print(f"Bernoulli NB Accuracy   : {accuracy_score(y_test, y_pred_bnb):.4f}")
print(f"Bernoulli NB Macro F1   : {f1_score(y_test, y_pred_bnb, average='macro'):.4f}")
```
